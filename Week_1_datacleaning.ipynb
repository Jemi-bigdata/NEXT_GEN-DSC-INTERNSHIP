OBSERVATION: The data set has 80 rows, 7 columns. Applicant's name has 1 missing value, email has 1 missing value, 
university has 2 values, and status has 1 missing value

CLEANING THE DATA

```python
#to find out the type of missing data I am working with
print(df[df['Applicant Name'].isna() == False]['Applicant Name'].unique())
```

```python
#turn empty strings into nan
#drop rows with no name
df['Applicant Name'] = df['Applicant Name'].replace('   ', np.nan)
df = df.dropna(subset=['Applicant Name'])
```

JUSTIFICATION: A record with no name lacks critical information needed to
identify the applicant

```python
#fill rows with missing data with placeholder values
df['Email'] = df['Email'].fillna('None@example.com')
df['University'] = df['University'].fillna('Not Specified')
df['Status'] = df['Status'].fillna('Under Review')
```

```python
df.duplicated().sum()
```

```python
#drop duplicates
df = df.drop_duplicates()
```

```python
#reset the index to avod inconsistencies in row labelling, since some rows were dropped
df = df.reset_index(drop=True)
```

```python
df.head()
```

STANDARDISING DATA

```python
#removing white spaces and standardise casing
df['Applicant Name'] = df['Applicant Name'].str.strip().str.title()
```

```python
df['Domain Applied'].unique()
```

```python
domain_mapping = {
 'web dev': 'Web Development',
 'WEB DEV': 'Web Development',
 'web development': 'Web Development',
 'data-sci': 'Data Science',
   'DATA SCI': 'Data Science'
 }
df['Domain Applied'] = df['Domain Applied'].replace(domain_mapping)
```

```python
df['Domain Applied'].unique()
```

```python
df['Status'].unique()
```

```python
status_mapping= {
    'selected': 'Selected'
}
df['Status']= df['Status'].replace(status_mapping)
```

```python
df['Status'].unique()
```

```python
df['Status'] = df['Status'].str.strip().str.title()
```

FIX DATA TYPES TO ENABLE PROPER CALCULATION ON THE DATA

```python
df['Application Date']= pd.to_datetime(df['Application Date'], format='mixed', errors='coerce')
```

```python
#phone number cleaning
df['Phone'] = df['Phone'].str.replace('-', '', regex=False).str.replace(' ', '', regex=False)
```
