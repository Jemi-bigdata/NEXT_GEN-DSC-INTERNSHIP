```python
import pandas as pd
import numpy as np
from faker import Faker
```

```python
# Initialize Faker for realistic mock names and emails
fake = Faker()
Faker.seed(42)
```

```python
data = []
domains = ['Web Development', 'Data Science', 'Mobile App Dev', 'Cybersecurity', 'UI/UX Design']
statuses = ['Under Review', 'Selected', 'Rejected']
```

```python
for i in range(80):
    name = fake.name()
    email = fake.email()
    phone = f"0{np.random.randint(200, 900)}-{np.random.randint(1000000, 9999999)}"
    domain = np.random.choice(domains)
    university = fake.company() + " University"
    # Mix date formats intentionally
    date = fake.date_between(start_date='-30d', end_date='today').strftime('%Y-%m-%d' if np.random.rand() > 0.5 else '%d/%m/%Y')
    status = np.random.choice(statuses)
    
    data.append([name, email, phone, domain, university, date, status])
```

```python
df = pd.DataFrame(data, columns=['Applicant Name', 'Email', 'Phone', 'Domain Applied', 'University', 'Application Date', 'Status'])
```

```python
# 1. Whitespace and missing values in Name
df.loc[2, 'Applicant Name'] = "   "
df.loc[15, 'Applicant Name'] = np.nan

# 2. Missing emails and uppercase variations
df.loc[5, 'Email'] = np.nan
df.loc[12, 'Email'] = "JANE.DOE@EXAMPLE.COM"

# 3. Messy domain variants (inconsistent casing and spellings)
df.loc[3, 'Domain Applied'] = 'web dev'
df.loc[18, 'Domain Applied'] = 'WEB DEV'
df.loc[25, 'Domain Applied'] = 'data-sci'
df.loc[33, 'Domain Applied'] = 'DATA SCI'

# 4. Missing universities
df.loc[7, 'University'] = np.nan
df.loc[22, 'University'] = np.nan

# 5. Missing status and inconsistent casing
df.loc[10, 'Status'] = np.nan
df.loc[28, 'Status'] = 'selected' 

# 6. Inject exact duplicate rows
df.loc[40] = df.loc[1].copy()
df.loc[41] = df.loc[1].copy()
```

```python
df.to_csv('applicants.csv', index=False)
print("applicants.csv generated successfully with 82 rows and built-in data issues!")
```
